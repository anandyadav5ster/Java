```python
def binary_search(lst,itemtosearch):
    start = 0
    high = len(lst)-1
    found = False
    while(start<=high and not found):
        mid = (start+high)//2
        if itemtosearch == lst[mid]:
            found = True
        else:
            if itemtosearch<lst[mid]:
                high = mid-1
            else:
                start = mid+1
    return found

lst = [1,2,17,8]
print(binary_search(lst,17))                    
```

```python
# find the common element in 2 list
def unique_element():
    s1 = input("Enter the first string ")
    s2 = input("Enter the second string")
    ls1 = set(s1)
    print(f'ls1 is {ls1}')
    ls2 = set(s2)
    print(f'ls2 is {ls2}')
    lst = ls1 & ls2
    print(''.join(lst))

```

```python
unique_element()
```

```python
def count_frequency():
    s = input("Enter the string ")
    d = {}
    for i in s:
        d[i] = d.get(i,0)+1
    print(d)    
```

```python
count_frequency()
```

```python
# print the maximum value
def count_maximumvalue_string():
    s=input("Enter the string ")
    d = {}
    for i in s:
        if i in d:
            d[i] = d.get(i,0)+1
        else:
            d[i] = 1
    # res = max(k for k,v in d.items() if v==max(d.values()))
    res = max(d,key=d.get)
    print(res)
    
```

```python
count_maximumvalue_string()
```

```python
def reverse_sentence():
    sen = 'I am in Bangalore'
    words = sen.split(' ')
    rev =''
    for word in words:
        rev = word+' '+ rev
    print(rev)
```

```python
reverse_sentence()
```

```python
def find_maximum_count_string():
    s = 'Anand'
    d = {}
    for i in s:
        d[i] = d.get(i,0)+1
    print(d)   
    res = [k for k,v in d.items() if v==max(d.values())]
    print(res)
```

```python
def count_word_sen():
    sen = "I am in Bangalore"
    words = sen.split(' ')
    d = {}
    for word in words:
        d[word] = d.get(word,len(word)+1)
    print(d) 
    res = max(d,key=d.get)
    print(res)
    item = max(d.items(), key=lambda x:x[1])
    print(item)
```
