```python
# Reverse a string

def reverse_string(s):
    res =''
    for i in s:
        res = i+res
    print(res)
```

```python
reverse_string('Anand')
```

```python
s= 'Bangalore'
res = s[::-1]
print(res)
```

```python
# Reverse sentence
def reverse_sentence(sen):
    words = sen.split(' ')[::-1]
    rev =[]
    print(words)
    for word in words:
        rev.append(word)
    # print(rev)  
    print(' '.join(rev))
    
        
```

```python
 reverse_sentence('Bangalore is good city')
```

```python
#  count words in string
def count_words(s):
    d ={}
    for i in s:
        if i in d:
            d[i] = d.get(i,0)+1
        else:
            d[i] =1
    print(d)        
        
```

```python
count_words('Bangalore')
```

```python
# remove duplicate from string
def remove_duplicate(s):
    res = set(s)
    print(res)
remove_duplicate('Anand')    
```

```python
def remove_duplicate_sen(sen):
    sen = sen.split( )
    res = set(sen)
    print(res)
    print(' '.join(res))
```
