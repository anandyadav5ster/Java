```python
# Reverse a string

def reverse_string(s):
    res =''
    for i in s:
        res = i+res
    print(res)
```

```python
reverse_string('Anand')
```

```python
# Reverse sentence
def reverse_sentence(sen):
    words = sen.split(' ')[::-1]
    rev =[]
    print(words)
    for word in words:
        rev.append(word)
    # print(rev)  
    print(' '.join(rev))
    
        
```

```python
 reverse_sentence('Bangalore is good city')
```

```python
#  count words in string
def count_words(s):
    d ={}
    for i in s:
        if i in d:
            d[i] = d.get(i,0)+1
        else:
            d[i] =1
    print(d)        
        
```

```python
count_words('Bangalore')
```

```python
def count_maximum_string(sen):
    words = sen.split(' ')
    d = {}
    for i in words:
        if i in d:
            d[i] = d.get(i,0)+1
        else:
            d[i] =1
    print(d)    
    # res = [k for k,v in d.items() if v==max(d,d.values())]
    res = max(d,key=d.get)
    print(res)
        
```

```python
count_maximum_string('I am am Anand')
```

```python
def count_maximum_lenth_string(sen):
    words = sen.split(' ')
    longest_word = max(words,key=len)
    print(longest_word)
```

```python
count_maximum_lenth_string('I am am Anand')
```
